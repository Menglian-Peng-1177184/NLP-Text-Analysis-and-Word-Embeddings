#Bonus task

```python
import string
articles = [article1.cleaned_text, article2.cleaned_text, article3.cleaned_text, article4.cleaned_text]

def clean_text(articles):
    articles = str(articles)
    articles = articles.lower()
    articles = articles.translate(str.maketrans('', '', string.punctuation))
    articles = ' '.join(articles.split())
    return articles

articles_cleaned = [clean_text(article) for article in articles]

words = []
for text in articles_cleaned:
    for word in text.split(' '):
        if word:
          words.append(word)

words = set(words)
print(words)
```

#Extract the words - skipgrams

```python
word2int = {}

for i,word in enumerate(words) :
    word2int[word] = i

sentences = []
for sentence_text in articles_cleaned :
    sentences.append(sentence_text.split())
print(sentences)

WINDOW_SIZE = 2

data = []
for sentence in sentences :
    for idx,word in enumerate(sentence) :
        for neighbor in sentence[max(idx - WINDOW_SIZE, 0) : min(idx + WINDOW_SIZE, len(sentence)) + 1] :
            if neighbor != word :
                data.append([word, neighbor])
print(data)
```

```python
import pandas as pd

for text in articles:
    print(text)

df = pd.DataFrame(data, columns = ['input', 'label'])
print(df)
```

#Define a Tensor flow model

```python
import tensorflow.compat.v1 as tf

tf.compat.v1.disable_eager_execution()
import numpy as np

ONE_HOT_DIM = len(words)

def to_one_hot_encoding(data_point_index) :
    one_hot_encoding = np.zeros(ONE_HOT_DIM)
    one_hot_encoding[data_point_index] = 1
    return one_hot_encoding

X = []
Y = []

for x, y in zip(df['input'], df['label']) :
    X.append(to_one_hot_encoding(word2int[ x ]))
    Y.append(to_one_hot_encoding(word2int[ y ]))

X_train = np.asarray(X)
Y_train = np.asarray(Y)

x = tf.placeholder(tf.float32, shape=(None, ONE_HOT_DIM))
y_label = tf.placeholder(tf.float32, shape=(None, ONE_HOT_DIM))

EMBEDDING_DIM = 2

W1 = tf.Variable(tf.random_normal([ONE_HOT_DIM, EMBEDDING_DIM]))
b1 = tf.Variable(tf.random_normal([EMBEDDING_DIM]))
hidden_layer = tf.add(tf.matmul(x,W1), b1)

W2 = tf.Variable(tf.random_normal([EMBEDDING_DIM, ONE_HOT_DIM]))
b2 = tf.Variable(tf.random_normal([ONE_HOT_DIM]))
prediction = tf.nn.softmax(tf.add( tf.matmul(hidden_layer, W2), b2))

loss = tf.reduce_mean(-tf.reduce_sum(y_label * tf.log(prediction), axis=[1]))

train_op = tf.train.GradientDescentOptimizer(0.05).minimize(loss)
```

#Train the model

```python
sess = tf.Session()
init = tf.global_variables_initializer()
sess.run(init)

iteration = 30000

for i in range(iteration) :
    # input : X_train(one-hot encoded word)
    # label : Y_train(one-hot encoded neighbor word)
    sess.run(train_op, feed_dict={x: X_train, y_label: Y_train})

    if i % 3000 == 0 :
        print('iteration ' + str(i) + ' loss is : ', sess.run(loss, feed_dict={x: X_train, y_label: Y_train}))
```

**2D coordinates for the words**

```python
vectors = sess.run(W1)
print(vectors)
```
